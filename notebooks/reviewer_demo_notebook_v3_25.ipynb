```python
from __future__ import annotations

from pathlib import Path
import os, sys, subprocess, shutil, textwrap
import json
import numpy as np

def find_repo_root(start: Path | None = None) -> Path:
    """Find repo root by walking up until we see required folders."""
    if start is None:
        start = Path.cwd().resolve()
    else:
        start = start.resolve()
    for p in [start] + list(start.parents):
        if (p / "configs").is_dir() and (p / "models").is_dir():
            return p
    raise RuntimeError("Could not find repo root (expected folders: configs/, models/).")

REPO = find_repo_root()
print("CWD :", Path.cwd())
print("REPO:", REPO)

# Make all relative paths resolve from repo root
os.chdir(REPO)
print("CHDIR ->", Path.cwd())

def R(rel: str | Path) -> Path:
    return (REPO / Path(rel)).resolve()

def ensure_dir(p: Path) -> None:
    p.mkdir(parents=True, exist_ok=True)

def run(cmd, env=None, cwd=None):
    """Run a command, printing stdout/stderr. Raises on failure."""
    cmd = [str(x) for x in cmd]
    print(">>", " ".join(cmd))
    r = subprocess.run(cmd, cwd=str(cwd or REPO), env=env, capture_output=True, text=True)
    if r.stdout:
        print(r.stdout)
    if r.returncode != 0:
        if r.stderr:
            print("----- STDERR -----")
            print(r.stderr)
        raise RuntimeError(f"Command failed (code={r.returncode})")
    return r

# Ensure imports like `import src...` work when scripts are executed via subprocess
BASE_ENV = os.environ.copy()
BASE_ENV["PYTHONPATH"] = str(REPO)
```

```python
def migrate_dataset_if_found(folder_name: str):
    wrong = REPO / folder_name
    correct = REPO / "data" / folder_name
    if wrong.exists() and wrong.is_dir():
        ensure_dir(correct.parent)
        if correct.exists():
            # If target already populated, keep it.
            if any(correct.iterdir()):
                print("[MIGRATE] target already populated:", correct)
                return
            else:
                shutil.rmtree(correct)
        print("[MIGRATE] moving", wrong, "->", correct)
        shutil.move(str(wrong), str(correct))

migrate_dataset_if_found("taylor_data_physchem_v4_deg25")
migrate_dataset_if_found("taylor_data_physchem_v4_interval")

print("[CHECK]")
print(" root dir exists   :", OUT_ROOT_DIR.exists())
print(" interval dir exists:", OUT_INTERVAL_DIR.exists())
```

## 7) Evaluation (K-sweep + baseline + reports/plots)

This uses a YAML config file and writes everything to `results/`.
You need trained checkpoints for the models you want to include.

Uncomment to run evaluation after training.


```python
# [ipynb] run_evaluation() for multiple degrees

import os, sys, subprocess, copy, yaml, torch
from pathlib import Path
import numpy as np

def build_truncated_test_npz(master_npz_path: str, degree: int, out_npz_path: str):
    master = np.load(master_npz_path, allow_pickle=True)

    coeffs = master["coeffs"]
    need_dim = degree + 1

    if coeffs.shape[1] < need_dim:
        raise ValueError(
            f"master coeff dim too small: got {coeffs.shape[1]}, need {need_dim}"
        )

    payload = {}
    for k in master.files:
        if k == "coeffs":
            payload[k] = coeffs[:, :need_dim].astype(np.float32, copy=False)
        elif k == "degree":
            payload[k] = np.full((coeffs.shape[0],), degree, dtype=np.int32)
        else:
            payload[k] = master[k]

    out_path = Path(out_npz_path)
    out_path.parent.mkdir(parents=True, exist_ok=True)
    np.savez_compressed(out_path, **payload)
    return out_path

def run_evaluation_multi(
    degrees=(15,),
    save_log=True,
    open_less=False,
):
    script = R("scripts/eval/evaluate_k_sweep.py")
    base_cfg_path = R("configs/eval_k_sweep.yaml")
    out_root = R("results/runs_k_sweep_viz_multi")
    ensure_dir(out_root)

    # master dataset: deg25
    master_test_npz = Path(
        "/home/seokjun/taylor-root-prediction/data/test_multi/"
        "taylor_test_physchem_v3_allroots_deg50_10000.npz"
    )

    with open(base_cfg_path, "r", encoding="utf-8") as f:
        base_cfg = yaml.safe_load(f)

    print("eval script =", script)
    print("master test =", master_test_npz)

    for degree in degrees:
        cfg = copy.deepcopy(base_cfg)

        # degree별 잘린 임시 test npz 생성
        truncated_test_npz = Path(
            f"/home/seokjun/taylor-root-prediction/data/test_multi_truncated/"
            f"taylor_test_physchem_v3_allroots_deg25_to_deg{degree}_10000.npz"
        )

        build_truncated_test_npz(
            master_npz_path=str(master_test_npz),
            degree=degree,
            out_npz_path=str(truncated_test_npz),
        )

        cfg["data"]["test_npz"] = str(truncated_test_npz)

        # 혹시 남아있으면 제거
        if "eval_degree" in cfg.get("data", {}):
            cfg["data"].pop("eval_degree", None)
        if "test_npz_pattern" in cfg.get("data", {}):
            cfg["data"].pop("test_npz_pattern", None)

        cfg["models"]["ast_ckpt"] = "/home/seokjun/math_12_3/expr_center_ast_best_50.pt"
        cfg["models"]["anchored_ckpt"] = f"/home/seokjun/taylor-root-prediction/results/taylor_nn/mlp/deg{degree}/best.pt"
        cfg["models"]["ann_ckpt"] = f"/home/seokjun/taylor-root-prediction/results/taylor_nn/ann/deg{degree}/best.pt"
        cfg["models"]["lstm_ckpt"] = f"/home/seokjun/taylor-root-prediction/results/taylor_nn/lstm/deg{degree}/best.pt"

        outdir = out_root / f"deg{degree}"
        ensure_dir(outdir)

        tmp_cfg = outdir / f"eval_k_sweep_deg{degree}.yaml"
        with open(tmp_cfg, "w", encoding="utf-8") as f:
            yaml.safe_dump(cfg, f, allow_unicode=True, sort_keys=False)

        env = BASE_ENV.copy()
        env.update({
            "EVAL_CFG": str(tmp_cfg),
            "OUTDIR": str(outdir),
            "DEVICE": ("cuda" if torch.cuda.is_available() else "cpu"),
        })

        print(f"\n[RUN EVAL] degree={degree}")
        print("test_npz =", cfg["data"]["test_npz"])
        print("ann_ckpt =", cfg["models"]["ann_ckpt"])
        print("mlp_ckpt =", cfg["models"]["anchored_ckpt"])
        print("lstm_ckpt =", cfg["models"]["lstm_ckpt"])

        if save_log:
            log_dir = R("results/logs")
            ensure_dir(log_dir)
            log_path = Path(log_dir) / f"eval_k_sweep_deg{degree}.log"

            with open(log_path, "w", encoding="utf-8") as f:
                p = subprocess.Popen(
                    [sys.executable, str(script)],
                    cwd=str(REPO),
                    env=env,
                    stdout=f,
                    stderr=subprocess.STDOUT,
                    text=True,
                    bufsize=1,
                )
                ret = p.wait()

            print("return_code =", ret)
            print("log_saved_to =", log_path)

            if open_less:
                subprocess.run(["bash", "-lc", f"less -R {str(log_path)}"], cwd=str(REPO))
        else:
            run([sys.executable, str(script)], env=env, cwd=REPO)

# 실행 예시
run_evaluation_multi(degrees=[25], save_log=True, open_less=False)
```

```python
import json
import re
import math
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# --------------------------------------------------
# 경로 유틸
# --------------------------------------------------
# R(), REPO 가 기존 노트북/프로젝트에서 이미 정의돼 있지 않다면
# 아래 두 줄을 환경에 맞게 수정하세요.
REPO = Path(".").resolve()

def R(rel_path: str) -> Path:
    return REPO / rel_path

# --------------------------------------------------
# 수식 문자열 정리
# --------------------------------------------------
def sanitize_expr_for_eval(raw: str) -> str:
    s = str(raw).strip()

    if "= 0" in s:
        s = s.split("= 0")[0].strip()
    elif "=0" in s:
        s = s.split("=0")[0].strip()

    # 맨 끝의 괄호 설명 제거
    s = re.sub(r"\s*\([^()]*\)\s*$", "", s).strip()

    # 거듭제곱 표기 변환
    s = s.replace("^", "**")

    # 혹시 이미 np.가 붙어 있는 경우 중복 방지용으로 잠깐 제거
    s = re.sub(r"\bnp\.", "", s)

    # 긴 이름 먼저 치환해야 log -> np.log 가 log10을 깨지 않음
    replacements = [
        ("log10", "np.log10"),
        ("sqrt", "np.sqrt"),
        ("tanh", "np.tanh"),
        ("sinh", "np.sinh"),
        ("cosh", "np.cosh"),
        ("exp", "np.exp"),
        ("sin", "np.sin"),
        ("cos", "np.cos"),
        ("tan", "np.tan"),
        ("log", "np.log"),
        ("abs", "np.abs"),
    ]

    for src, dst in replacements:
        s = re.sub(rf"(?<!np\.)\b{src}\s*\(", f"{dst}(", s)

    s = re.sub(r"\bln\s*\(", "np.log(", s)

    return s

# --------------------------------------------------
# eval callable 생성
# --------------------------------------------------
def make_callable(expr_sanitized: str):
    code = compile(expr_sanitized, "<expr>", "eval")
    allowed = {"np": np, "math": math}

    def f(x):
        with np.errstate(all="ignore"):
            return eval(code, {"__builtins__": {}}, {**allowed, "x": x})

    return f

# --------------------------------------------------
# 메인 실행
# --------------------------------------------------
outdir = R("results/runs_k_sweep_viz")
jsons = sorted(outdir.rglob("fail_by_funcid_*.json"))

if not jsons:
    print("[INFO] No fail_by_funcid_*.json found. Turn on reports.report_fail_save: true and rerun evaluation.")
else:
    path = jsons[0]
    print("Using:", path.relative_to(REPO))

    rows = json.loads(path.read_text(encoding="utf-8"))

    # 저장 폴더
    save_dir = outdir / "failure_examples"
    save_dir.mkdir(parents=True, exist_ok=True)
    print("Save dir:", save_dir)

    # 필요하면 일부만 쓰세요. 예: rows[:5]
    for r in rows[:]:
        fid = r.get("func_id")
        n_fail = r.get("n_fail")
        n_all = r.get("n_all")
        exs = r.get("expr_examples", [])[:2]   # expr 2개만 사용

        if not exs:
            print(f"[SKIP] func_id={fid}: no expr_examples")
            continue

        xs = np.linspace(-20, 20, 600)

        fig, ax = plt.subplots(figsize=(10, 4))
        plotted = 0

        for i, e in enumerate(exs):
            san = sanitize_expr_for_eval(e)
            try:
                f = make_callable(san)
                ys = f(xs)
                ys = np.asarray(ys, dtype=float)

                # 유효하지 않은 값 처리
                ys = np.where(np.isfinite(ys), ys, np.nan)

                ax.plot(
                    xs,
                    np.log10(np.abs(ys) + 1e-30),
                    label=f"expr#{i+1}"
                )
                plotted += 1

            except Exception as ex:
                print(f"[WARN] func_id={fid}, expr#{i+1} plot failed: {ex}")
                continue

        if plotted == 0:
            plt.close(fig)
            print(f"[SKIP] func_id={fid}: no valid plot")
            continue

        ax.set_title(f"Fail examples: func_id={fid} | n_fail={n_fail} / n_all={n_all}")
        ax.set_xlabel("x")
        ax.set_ylabel("log10(|f(x)| + eps)")
        ax.grid(True, alpha=0.25)
        ax.legend()

        plt.tight_layout()

        # ------------------------------------------
        # 저장: PNG + PDF + SVG
        # PNG  : 래스터
        # PDF  : 벡터
        # SVG  : 벡터
        # ------------------------------------------
        png_path = save_dir / f"fail_example_funcid_{fid}.png"
        pdf_path = save_dir / f"fail_example_funcid_{fid}.pdf"
        svg_path = save_dir / f"fail_example_funcid_{fid}.svg"

        fig.savefig(png_path, dpi=300, bbox_inches="tight")
        fig.savefig(pdf_path, bbox_inches="tight")
        fig.savefig(svg_path, bbox_inches="tight")

        print("Saved PNG :", png_path)
        print("Saved PDF :", pdf_path)
        print("Saved SVG :", svg_path)

        plt.show()
        plt.close(fig)
```

```python
import json
import re
import math
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

REPO = Path(".").resolve()

def R(rel_path: str) -> Path:
    return REPO / rel_path

def sanitize_expr_for_eval(raw: str) -> str:
    s = str(raw).strip()

    if "= 0" in s:
        s = s.split("= 0")[0].strip()
    elif "=0" in s:
        s = s.split("=0")[0].strip()

    s = re.sub(r"\s*\([^()]*\)\s*$", "", s).strip()
    s = s.replace("^", "**")
    s = re.sub(r"\bnp\.", "", s)

    replacements = [
        ("log10", "np.log10"),
        ("sqrt", "np.sqrt"),
        ("tanh", "np.tanh"),
        ("sinh", "np.sinh"),
        ("cosh", "np.cosh"),
        ("exp", "np.exp"),
        ("sin", "np.sin"),
        ("cos", "np.cos"),
        ("tan", "np.tan"),
        ("log", "np.log"),
        ("abs", "np.abs"),
    ]

    for src, dst in replacements:
        s = re.sub(rf"(?<!np\.)\b{src}\s*\(", f"{dst}(", s)

    s = re.sub(r"\bln\s*\(", "np.log(", s)
    return s

def make_callable(expr_sanitized: str):
    code = compile(expr_sanitized, "<expr>", "eval")
    allowed = {"np": np, "math": math}

    def f(x):
        with np.errstate(all="ignore"):
            return eval(code, {"__builtins__": {}}, {**allowed, "x": x})

    return f

outdir = R("results/runs_k_sweep_viz")
jsons = sorted(outdir.rglob("fail_by_funcid_*.json"))

if not jsons:
    print("[INFO] No fail_by_funcid_*.json found.")
else:
    path = jsons[0]
    print("Using:", path.relative_to(REPO))

    rows = json.loads(path.read_text(encoding="utf-8"))

    save_dir = outdir / "failure_examples"
    save_dir.mkdir(parents=True, exist_ok=True)
    print("Save dir:", save_dir)

    xs = np.linspace(-20, 20, 600)

    for r in rows:
        fid = r.get("func_id")
        n_fail = r.get("n_fail", 0)
        n_all = r.get("n_all", 0)

        # JSON 키 이름은 실제 파일에 맞게 조정
        fail_expr = r.get("fail_expr")
        all_expr = r.get("all_expr")

        # 혹시 리스트 구조면 첫 번째 것 사용
        if fail_expr is None:
            fail_examples = r.get("fail_examples", [])
            if fail_examples:
                fail_expr = fail_examples[0]

        if all_expr is None:
            all_examples = r.get("all_examples", [])
            if all_examples:
                all_expr = all_examples[0]

        if fail_expr is None and all_expr is None:
            print(f"[SKIP] func_id={fid}: no fail/all expr")
            continue

        fig, ax = plt.subplots(figsize=(10, 4))
        plotted = 0

        plot_items = [
            ("fail", fail_expr, "tab:blue"),
            ("all",  all_expr,  "tab:orange"),
        ]

        for label, expr_raw, color in plot_items:
            if not expr_raw:
                continue

            try:
                san = sanitize_expr_for_eval(expr_raw)
                f = make_callable(san)
                ys = f(xs)
                ys = np.asarray(ys, dtype=float)
                ys = np.where(np.isfinite(ys), ys, np.nan)

                ax.plot(
                    xs,
                    np.log10(np.abs(ys) + 1e-30),
                    label=label,
                    color=color,
                    linewidth=2.0,
                )
                plotted += 1

            except Exception as ex:
                print(f"[WARN] func_id={fid}, {label} plot failed: {ex}")

        if plotted == 0:
            plt.close(fig)
            print(f"[SKIP] func_id={fid}: no valid plot")
            continue

        ax.set_title(f"func_id={fid} | fail={n_fail} / all={n_all}")
        ax.set_xlabel("x")
        ax.set_ylabel("log10(|f(x)| + eps)")
        ax.grid(True, alpha=0.25)
        ax.legend(title=None)

        plt.tight_layout()

        png_path = save_dir / f"fail_all_funcid_{fid}.png"
        pdf_path = save_dir / f"fail_all_funcid_{fid}.pdf"
        svg_path = save_dir / f"fail_all_funcid_{fid}.svg"

        fig.savefig(png_path, dpi=600, bbox_inches="tight")
        # fig.savefig(pdf_path, bbox_inches="tight")
        # fig.savefig(svg_path, bbox_inches="tight")

        print("Saved PNG :", png_path)
        # print("Saved PDF :", pdf_path)
        # print("Saved SVG :", svg_path)

        plt.show()
        plt.close(fig)
```

```python
import json
import re
from pathlib import Path

REPO = Path(".").resolve()

def R(rel_path: str) -> Path:
    return REPO / rel_path

# ------------------------------------------
# 입력: log txt 파일
# ------------------------------------------
log_path = R("results/logs/eval_k_sweep_deg25.txt")   # 실제 경로에 맞게 수정
text = log_path.read_text(encoding="utf-8")

# ------------------------------------------
# 1) FAIL 블록 파싱
# ------------------------------------------
def parse_fail_concentration_block(text: str):
    out = {}

    m = re.search(
        r"=+\s*ALL-METHOD FAIL CONCENTRATION.*?=+\s*\n(.*?)=+\s*END FAIL CONCENTRATION REPORT\s*=+",
        text,
        flags=re.S
    )
    if not m:
        return out

    block = m.group(1)

    entry_pat = re.compile(
        r"^\s*\d+\s*\|\s*(?P<func_id>\d+)\s*\|\s*(?P<n_fail>\d+)\s*/\s*(?P<n_all>\d+)\s*\|\s*(?P<fail_rate>[\d.]+)%\s*\|\s*(?P<reason>.+?)\n"
        r"(?:\s*expr1:\s*(?P<expr1>.+?)\n)?"
        r"(?:\s*expr2:\s*(?P<expr2>.+?)\n)?"
        r"(?:\s*expr3:\s*(?P<expr3>.+?)\n)?"
        r"(?=^\s*-{10,}|\Z)",
        flags=re.M | re.S
    )

    for mm in entry_pat.finditer(block):
        fid = int(mm.group("func_id"))
        n_fail = int(mm.group("n_fail"))
        n_all = int(mm.group("n_all"))
        fail_rate = float(mm.group("fail_rate")) / 100.0
        reason_text = mm.group("reason").strip()

        exprs = [
            mm.group("expr1"),
            mm.group("expr2"),
            mm.group("expr3"),
        ]
        exprs = [e.strip() for e in exprs if e and e.strip()]

        top_reasons = []
        for part in [x.strip() for x in reason_text.split(",") if x.strip()]:
            mr = re.match(r"(.+):(\d+)$", part)
            if mr:
                top_reasons.append([mr.group(1), int(mr.group(2))])
            else:
                top_reasons.append([part, None])

        out[fid] = {
            "func_id": fid,
            "n_all": n_all,
            "n_fail": n_fail,
            "fail_rate": fail_rate,
            "top_reasons": top_reasons,
            "fail_expr_examples": exprs,
        }

    return out

# ------------------------------------------
# 2) FUNC_ID WINNER RATIO 블록 파싱
#    여기서는 전체 대표 예시(all_expr_examples) 추출
# ------------------------------------------
def parse_func_ratio_block(text: str):
    out = {}

    m = re.search(
        r"=+\s*FUNC_ID WINNER RATIO .*?=+\s*\n(.*)$",
        text,
        flags=re.S
    )
    if not m:
        return out

    block = m.group(1)

    entry_pat = re.compile(
        r"func_id=\s*(?P<func_id>\d+)\s*\|\s*n=\s*(?P<n_all>\d+)\s*\|.*?\n"
        r"\s*winner\(any\)\s*:\s*(?P<wany>.+?)\n"
        r"\s*winner\(okonly\)\s*:\s*(?P<wok>.+?)\n"
        r"(?:\s*expr1:\s*(?P<expr1>.+?)\n)?"
        r"(?:\s*expr2:\s*(?P<expr2>.+?)\n)?"
        r"(?:\s*expr3:\s*(?P<expr3>.+?)\n)?"
        r"(?=^\s*-{10,}|\Z)",
        flags=re.M | re.S
    )

    for mm in entry_pat.finditer(block):
        fid = int(mm.group("func_id"))
        n_all = int(mm.group("n_all"))

        exprs = [
            mm.group("expr1"),
            mm.group("expr2"),
            mm.group("expr3"),
        ]
        exprs = [e.strip() for e in exprs if e and e.strip()]

        out[fid] = {
            "func_id": fid,
            "n_all": n_all,
            "all_expr_examples": exprs,
        }

    return out

fail_data = parse_fail_concentration_block(text)
ratio_data = parse_func_ratio_block(text)

# ------------------------------------------
# 3) 병합
# ------------------------------------------
all_func_ids = sorted(set(fail_data.keys()) | set(ratio_data.keys()))
merged = []

for fid in all_func_ids:
    row = {
        "func_id": fid,
        "n_all": None,
        "n_fail": 0,
        "fail_rate": 0.0,
        "top_reasons": [],
        "fail_expr_examples": [],
        "all_expr_examples": [],
    }

    if fid in ratio_data:
        row["n_all"] = ratio_data[fid]["n_all"]
        row["all_expr_examples"] = ratio_data[fid]["all_expr_examples"]

    if fid in fail_data:
        row["n_all"] = fail_data[fid]["n_all"]
        row["n_fail"] = fail_data[fid]["n_fail"]
        row["fail_rate"] = fail_data[fid]["fail_rate"]
        row["top_reasons"] = fail_data[fid]["top_reasons"]
        row["fail_expr_examples"] = fail_data[fid]["fail_expr_examples"]

    merged.append(row)

# ------------------------------------------
# 4) 확인 출력
# ------------------------------------------
print(json.dumps(merged, indent=2, ensure_ascii=False))

# 필요하면 저장
out_path = R("representative_fail_all_from_log.json")
out_path.write_text(json.dumps(merged, indent=2, ensure_ascii=False), encoding="utf-8")
print(f"saved: {out_path}")
```

```python
import json
import re
import math
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

REPO = Path(".").resolve()

def R(rel_path: str) -> Path:
    return REPO / rel_path
FUNC_NAME_MAP = {
    0: "Damped Oscillator 1",
    1: "Damped Oscillator 2",
    2: "Asymmetric Double-Well Potential",
    3: "Spring-Tah",
    5: "Damped Sine Current",
    6: "Heat-Fin-like",
    8: "Ising Magnetization",
    9: "Activation + Logarithmic Term",
    10: "Arrhenius-like",
    11: "Diode I-V",
    12: "Damped Wave",
    13: "Reaction Coordinate",
    14: "Michaelis-Menten",
}
# --------------------------------------------------
# 설정
# --------------------------------------------------
log_path = R("results/logs/eval_k_sweep_deg25.txt")   # 실제 경로에 맞게 수정
save_dir = R("results/runs_k_sweep_viz/fail_all_plots_from_log")
save_dir.mkdir(parents=True, exist_ok=True)

combined_png_path = save_dir / "combined_fail_all_plots.png"
combined_json_path = save_dir / "representative_fail_all_from_log.json"

# --------------------------------------------------
# 수식 문자열 정리
# --------------------------------------------------
def sanitize_expr_for_eval(raw: str) -> str:
    s = str(raw).strip()

    if "= 0" in s:
        s = s.split("= 0")[0].strip()
    elif "=0" in s:
        s = s.split("=0")[0].strip()

    s = re.sub(r"\s*\([^()]*\)\s*$", "", s).strip()
    s = s.replace("^", "**")
    s = re.sub(r"\bnp\.", "", s)

    replacements = [
        ("log10", "np.log10"),
        ("sqrt", "np.sqrt"),
        ("tanh", "np.tanh"),
        ("sinh", "np.sinh"),
        ("cosh", "np.cosh"),
        ("exp", "np.exp"),
        ("sin", "np.sin"),
        ("cos", "np.cos"),
        ("tan", "np.tan"),
        ("log", "np.log"),
        ("abs", "np.abs"),
    ]

    for src, dst in replacements:
        s = re.sub(rf"(?<!np\.)\b{src}\s*\(", f"{dst}(", s)

    s = re.sub(r"\bln\s*\(", "np.log(", s)
    return s

def make_callable(expr_sanitized: str):
    code = compile(expr_sanitized, "<expr>", "eval")
    allowed = {"np": np, "math": math}

    def f(x):
        with np.errstate(all="ignore"):
            return eval(code, {"__builtins__": {}}, {**allowed, "x": x})

    return f

# --------------------------------------------------
# 로그 파싱
# --------------------------------------------------
def parse_fail_concentration_block(text: str):
    out = {}

    m = re.search(
        r"=+\s*ALL-METHOD FAIL CONCENTRATION.*?=+\s*\n(.*?)=+\s*END FAIL CONCENTRATION REPORT\s*=+",
        text,
        flags=re.S
    )
    if not m:
        return out

    block = m.group(1)

    entry_pat = re.compile(
        r"^\s*\d+\s*\|\s*(?P<func_id>\d+)\s*\|\s*(?P<n_fail>\d+)\s*/\s*(?P<n_all>\d+)\s*\|\s*(?P<fail_rate>[\d.]+)%\s*\|\s*(?P<reason>.+?)\n"
        r"(?:\s*expr1:\s*(?P<expr1>.+?)\n)?"
        r"(?:\s*expr2:\s*(?P<expr2>.+?)\n)?"
        r"(?:\s*expr3:\s*(?P<expr3>.+?)\n)?"
        r"(?=^\s*-{10,}|\Z)",
        flags=re.M | re.S
    )

    for mm in entry_pat.finditer(block):
        fid = int(mm.group("func_id"))
        n_fail = int(mm.group("n_fail"))
        n_all = int(mm.group("n_all"))
        fail_rate = float(mm.group("fail_rate")) / 100.0
        reason_text = mm.group("reason").strip()

        exprs = [mm.group("expr1"), mm.group("expr2"), mm.group("expr3")]
        exprs = [e.strip() for e in exprs if e and e.strip()]

        top_reasons = []
        for part in [x.strip() for x in reason_text.split(",") if x.strip()]:
            mr = re.match(r"(.+):(\d+)$", part)
            if mr:
                top_reasons.append([mr.group(1), int(mr.group(2))])
            else:
                top_reasons.append([part, None])

        out[fid] = {
            "func_id": fid,
            "n_all": n_all,
            "n_fail": n_fail,
            "fail_rate": fail_rate,
            "top_reasons": top_reasons,
            "fail_expr_examples": exprs,
        }

    return out

def parse_func_ratio_block(text: str):
    out = {}

    m = re.search(
        r"=+\s*FUNC_ID WINNER RATIO .*?=+\s*\n(.*)$",
        text,
        flags=re.S
    )
    if not m:
        return out

    block = m.group(1)

    entry_pat = re.compile(
        r"func_id=\s*(?P<func_id>\d+)\s*\|\s*n=\s*(?P<n_all>\d+)\s*\|.*?\n"
        r"\s*winner\(any\)\s*:\s*(?P<wany>.+?)\n"
        r"\s*winner\(okonly\)\s*:\s*(?P<wok>.+?)\n"
        r"(?:\s*expr1:\s*(?P<expr1>.+?)\n)?"
        r"(?:\s*expr2:\s*(?P<expr2>.+?)\n)?"
        r"(?:\s*expr3:\s*(?P<expr3>.+?)\n)?"
        r"(?=^\s*-{10,}|\Z)",
        flags=re.M | re.S
    )

    for mm in entry_pat.finditer(block):
        fid = int(mm.group("func_id"))
        n_all = int(mm.group("n_all"))

        exprs = [mm.group("expr1"), mm.group("expr2"), mm.group("expr3")]
        exprs = [e.strip() for e in exprs if e and e.strip()]

        out[fid] = {
            "func_id": fid,
            "n_all": n_all,
            "all_expr_examples": exprs,
        }

    return out

def build_merged_from_log(log_text: str):
    fail_data = parse_fail_concentration_block(log_text)
    ratio_data = parse_func_ratio_block(log_text)

    all_func_ids = sorted(set(fail_data.keys()) | set(ratio_data.keys()))
    merged = []

    for fid in all_func_ids:
        row = {
            "func_id": fid,
            "n_all": None,
            "n_fail": 0,
            "fail_rate": 0.0,
            "top_reasons": [],
            "fail_expr_examples": [],
            "all_expr_examples": [],
        }

        if fid in ratio_data:
            row["n_all"] = ratio_data[fid]["n_all"]
            row["all_expr_examples"] = ratio_data[fid]["all_expr_examples"]

        if fid in fail_data:
            row["n_all"] = fail_data[fid]["n_all"]
            row["n_fail"] = fail_data[fid]["n_fail"]
            row["fail_rate"] = fail_data[fid]["fail_rate"]
            row["top_reasons"] = fail_data[fid]["top_reasons"]
            row["fail_expr_examples"] = fail_data[fid]["fail_expr_examples"]

        merged.append(row)

    return merged

# --------------------------------------------------
# 대표식 선택
# --------------------------------------------------
def pick_representative_expr(expr_list):
    if not expr_list:
        return None
    return expr_list[0]

# --------------------------------------------------
# 단일 subplot에 그리기
# --------------------------------------------------
def draw_fail_all_on_ax(ax, row, x_min=-20, x_max=20, n=1200):
    fid = row["func_id"]
    n_fail = row["n_fail"]
    n_all = row["n_all"]

    fail_expr = pick_representative_expr(row.get("fail_expr_examples", []))
    all_expr = pick_representative_expr(row.get("all_expr_examples", []))

    xs = np.linspace(x_min, x_max, n)
    plotted = 0

    for label, expr_raw, color in [
        ("fail", fail_expr, "tab:blue"),
        ("all", all_expr, "tab:orange"),
    ]:
        if not expr_raw:
            continue

        try:
            san = sanitize_expr_for_eval(expr_raw)
            f = make_callable(san)
            ys = np.asarray(f(xs), dtype=float)
            ys = np.where(np.isfinite(ys), ys, np.nan)

            ax.plot(
                xs,
                np.log10(np.abs(ys) + 1e-30),
                label=label,
                color=color,
                linewidth=1.5,
            )
            plotted += 1
        except Exception as ex:
            print(f"[WARN] func_id={fid}, {label} plot failed: {ex}")

    if plotted == 0:
        ax.set_visible(False)
        return False

    func_name = FUNC_NAME_MAP.get(fid, f"func_id={fid}")
    ax.set_title(f"{func_name} | fail={n_fail} / all={n_all}", fontsize=8, pad=4)
    ax.set_xlabel("x", fontsize=8, labelpad=1)
    ax.set_ylabel("log10(|f(x)| + eps)", fontsize=8, labelpad=1)

    # tick 위치도 단순화
    ax.set_xticks([-20, 0, 20])

    # 숫자 글씨 크게 줄이기
    ax.tick_params(axis="x", labelsize=8, pad=1)
    ax.tick_params(axis="y", labelsize=8, pad=1)

    # 범례도 축소
    ax.legend(fontsize=5, loc="best", frameon=True,
              borderpad=0.2, handlelength=1.2, labelspacing=0.2)

    ax.grid(True, alpha=0.25)
    return True
# --------------------------------------------------
# 실행
# --------------------------------------------------
log_text = log_path.read_text(encoding="utf-8")
rows = build_merged_from_log(log_text)

# 중간 JSON 저장
combined_json_path.write_text(
    json.dumps(rows, indent=2, ensure_ascii=False),
    encoding="utf-8"
)
print(f"Saved merged JSON: {combined_json_path}")

# 1) n_fail == 0 제외
rows_to_plot = [row for row in rows if int(row.get("n_fail", 0)) > 0]

if not rows_to_plot:
    print("[INFO] No rows with n_fail > 0")
else:
    # 2) 모든 그림을 하나의 PNG로 합치기
    n_plots = len(rows_to_plot)
    ncols = 2
    nrows = math.ceil(n_plots / ncols)

    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=(14, 3.8 * nrows),
        squeeze=False
    )

    axes_flat = axes.flatten()

    used = 0
    for ax, row in zip(axes_flat, rows_to_plot):
        ok = draw_fail_all_on_ax(ax, row)
        if ok:
            used += 1

    # 남는 축 숨김
    for ax in axes_flat[len(rows_to_plot):]:
        ax.set_visible(False)

    fig.suptitle("Representative fail vs all curves", fontsize=16, y=0.995)
    plt.tight_layout(rect=[0, 0, 1, 0.985])

    fig.savefig(combined_png_path, dpi=600, bbox_inches="tight")
    print(f"Saved combined PNG (600 dpi): {combined_png_path}")

    plt.show()
    plt.close(fig)
```
